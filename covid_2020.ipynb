```python
import numpy as np
from scipy.optimize import curve_fit

# Filter data for Italy
italy_data = df2[df2['location'] == 'Italy']

# Convert the 'date' column to datetime objects
italy_data['date'] = pd.to_datetime(italy_data['date'])

# Extract dates and total cases
dates = italy_data['date'].values.astype(np.int64) // 10**9  # Convert dates to numeric values
total_cases = italy_data['total_cases'].values

# Define the exponential function
def exponential_func(x, A, b):
    return A * np.exp(b * x)

# Fit the exponential function to the data
popt, pcov = curve_fit(exponential_func, dates, total_cases)

# Get the predicted value for 2020-03-20
date_20200320 = pd.to_datetime('2020-03-20').value // 10**9
predicted_cases = exponential_func(date_20200320, *popt)

# Get the actual value for 2020-03-20
actual_cases = italy_data[italy_data['date'] == '2020-03-20']['total_cases'].values[0]

# Calculate the difference
difference = predicted_cases - actual_cases

print(f"Difference between predicted and actual cases on 2020-03-20: {difference}")
```
